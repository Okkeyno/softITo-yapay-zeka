```python
# 1. Değişkenlerle Basit Hesaplamalar
""" Bir ürünün birim fiyatını ve adedini değişkenlerde tutup toplam tutarı hesaplayın. """
birim_fiyat = 10.99
adet = 5
toplam_fiyat = birim_fiyat * adet
print(f"Toplam fiyat: {toplam_fiyat} TL")

"""maas adlı bir değişkene 8000 değeri verin, sonra += kullanarak 1500 TL zam ekleyip yeni maaşı yazdırın."""
maas = 8000
maas += 1500
print(f"Yeni maaş: {maas} TL")

# 2. İç İçe (Nested) if-else
"""Kullanıcıdan yaş ve öğrenci olup olmadığı bilgisini alıp; 65 yaş üstü ise "Ücretsiz",
öğrenci ise "İndirimli", diğer durumda "Tam ücret" yazdıran bir program yazın. """
yas = int(input("Yaşınızı girin: "))
ogrenci_mi = input("Öğrenci misiniz? (E/H): ").upper() == "E"
if yas >= 65:
  print("Ücretsiz")
elif ogrenci_mi:
  print("İndirimli")
else:
  print("Tam ücret")

"""Bir sayının hem pozitif hem de çift olup olmadığını iç içe if ile kontrol edin."""
sayi = int(input("Bir sayı girin: "))
if sayi > 0:
  if sayi % 2 == 0:
    print("Sayı pozitif ve çifttir.")
  else:
    print("Sayı pozitif ancak çift değildir.")
else:
  print("Sayı negatif veya sıfırdır.")
```

```python
# 3. elif ile Çoklu Koşullar
"""Kullanıcıdan sıcaklık değeri alıp; 30 üstü "Sıcak", 15-30 arası "Ilıman", 15 altı "Soğuk" yazdıran bir program yazın."""
sicaklik = int(input("Sıcaklık değerini girin: "))
if sicaklik > 30:
  print("Sıcak")
elif sicaklik >= 15:
  print("Ilıman")
else:
  print("Soğuk")

"""Kullanıcıdan bir ay numarası (1-12) alıp hangi mevsime ait olduğunu elif ile bulun."""
ay_numarasi = int(input("Bir ay numarası girin (1-12): "))
if ay_numarasi in [12, 1, 2]:
  print("Kış")
elif ay_numarasi in [3, 4, 5]:
  print("İlkbahar")
elif ay_numarasi in [6, 7, 8]:
  print("Yaz")
elif ay_numarasi in [9, 10, 11]:
  print("Sonbahar")
else:
  print("Geçersiz ay numarası.")
```

```python
# 4. for Döngüsü ile Liste Üzerinde İşlem
 """[5, 15, 8, 22, 3, 19] listesindeki en büyük sayıyı döngüyle bulun (max() kullanmadan). """
 liste = [5, 15, 8, 22, 3, 19]
 en_buyuk = liste[0]
 for sayi in liste:
   if sayi > en_buyuk:
     en_buyuk = sayi
 print(f"Listedeki en büyük sayı: {en_buyuk}")

 """Bir liste içindeki tüm sayıların kaç tanesinin 10'dan büyük olduğunu sayan bir kod yazın."""
 liste = [5, 15, 8, 22, 3, 19]
 sayac = 0
 for sayi in liste:
   if sayi > 10:
     sayac += 1
```

```python
# 5. while Döngüsü ile Basit Sayaç
"""10'dan 1'e geri sayan bir while döngüsü yazın."""
sayac = 10
while sayac > 0:
  print(sayac)
  sayac -= 1

"""Kullanıcıdan şifre alıp doğru şifreyi girene kadar tekrar tekrar soran bir program yazın."""
hedef_sifre = input("ilk şifrenizi oluşturun: ")

while True:
    girilen_sifre = input("Şifrenizi girin: ")

    if girilen_sifre == hedef_sifre:
        print("Giriş başarılı!")
        break
    else:
        print("Hatalı şifre")
```

```python
# 6. Döngü İçinde Koşul Kullanma
"""1'den 20'ye kadar olan sayılardan sadece 3'e tam bölünenleri yazdırın. """
for i in range(21):
  if i % 3 == 0:
    print(i)



"""Bir liste içindeki sayılardan negatif olanları ayrı, pozitif olanları ayrı yazdırın."""
liste = [-5, 10, -15, 20, -30]
pozitif = []
negatif = []
for i in liste:
  if i > 0:
    pozitif.append(i)
  else:
    negatif.append(i)

print("Pozitifler")
for i in pozitif:
  print(f"- {i}")

print("Negatifler")
for i in negatif:
  print(f"- {i}")
```

```python
# 7. Basit Fonksiyon Tanımlama
"""Bir sayının küpünü (3. kuvvetini) alan kup_al adlı bir fonksiyon yazın. """
def kup(sayi):
  hesap = sayi ** 3
  return hesap

print(kup(5))

"""İki sayıyı parametre olarak alıp toplamını döndüren bir fonksiyon yazın."""
def toplam(a,b):
  return a + b

a = int(input("Bir sayı girin: "))
b = int(input("Bir sayı girin: "))
print("Toplam: ", toplam(a,b))
```

```python
# 8. Fonksiyon + Koşul Birlikte
"""Bir sayının pozitif, negatif veya sıfır olduğunu döndüren sayi_durumu adlı bir fonksiyon yazın."""
def sayi_durumu(sayi):
  if sayi > 0:
    return "Pozitif"
  elif sayi < 0:
    return "Negatif"
  else:
    return "Sıfır"

print(sayi_durumu(5))

"""Üç sayı arasından en büyüğünü bulup döndüren bir fonksiyon yazın."""
def enbuyuk(x, y, z):
  if x > y and x > z:
    return f"en büyük x: {x}"
  elif y > x and y > z:
    return f"en büyük y: {y}"
  else:
    return f"en büyük z: {z}"

x = int(input("x girin: "))
y = int(input("y girin: "))
z = int(input("z girin: "))

print(enbuyuk(x,y,z))
```

```python
# 9. Liste Oluşturma ve İşleme
"""Boş bir liste oluşturup kullanıcıdan 5 tane şehir adı alıp listeye ekleyin, sonra listeyi numaralandırarak yazdırın."""
sehirler = []

for i in range(5):
  sehir = input("Şehir girin: ")
  sehirler.append(sehir)

for i, sehir in enumerate(sehirler):
  print(f"{i+1}. Şehir: {sehir}")

"""Bir listeye eklenen ürünlerden istenilen birini remove() ile silin ve güncel listeyi gösterin."""
sehirler.remove("adana")
for i, sehir in enumerate(sehirler):
  print(f"{i+1}. Şehir: {sehir}")
```

```python
# 10. Kullanıcıdan Birden Fazla Veri Alma
"""Kullanıcıdan ad, soyad ve doğum yılını alıp yaşını hesaplayan (2026'ya göre) bir program yazın."""
ad = input("Adınızı girin: ")
soyad = input("Soyadınızı girin: ")
dogum_yili = int(input("Doğum yılınızı girin: "))
yas = 2026 - dogum_yili
print(f"{ad} {soyad} kişisi {yas} yaşındadır.")

"""Kullanıcıdan boy (cm) ve kilo bilgisi alıp basit BMI (vücut kitle indeksi) hesaplayan bir program yazın."""
boy = int(input: "boyunuzu girin: "))
kilo = int(input: "kilonuzu girin: "))
bmi = kilo / (boy ** 2)
print(f"BMI sonucu : {bmi}")

```

```python
# 11. Küçük Bir Uygulama — Not Ortalaması Hesaplama
"""Kullanıcıdan 4 dersin notunu alıp ortalamayı hesaplayan ve harf notuna (AA, BA, BB...) çeviren bir program yazın."""
notlar = []
toplam = 0
for i in range(4):
  notum = int(input(f"{i+1}. dersin notunu girin: "))
  notlar.append(notum)
  toplam += notum
ort = toplam / len(notlar)

if ort >= 90:
  harf = "AA"
elif ort >= 80:
  harf = "BA"
elif ort >= 70:
  harf = "BB"
elif ort >= 45:
  harf = "CC"
else:
  harf = "FF"

print("Notunuz: ", harf)


"""Girilen notlardan en yüksek ve en düşük notu bulup ekrana yazdırın."""
print(f"en kucuk not: {min(notlar)}")
print(f"en büyük not: {max(notlar)}")
```

```python
# 12. Küçük Bir Uygulama — Basit Sepet/Alışveriş
""" Basit Sepet/Alışveriş Sepete eklenen ürünlerden 100 TL üzeri olanları ayrıca listeleyen bir program yazın."""
urunler = []
fiyatlar = []

for i in range(5):
    urun_adi = input(f"{i+1}. ürünün adını girin: ")
    urun_fiyati = float(input(f"{urun_adi} ürününün fiyatını girin (TL): "))
    urunler.append(urun_adi)
    fiyatlar.append(urun_fiyati)

print("\n100 TL üzerindeki ürünler:")
for urun, fiyat in zip(urunler, fiyatlar):
    if fiyat > 100:
        print(f"- {urun}: {fiyat} TL")

"""Toplam tutar 500 TL üzerindeyse \"%10 indirim kazandınız\" mesajı gösteren bir alışveriş programı yazın."""
toplam_tutar = sum(fiyatlar)
print(f"\nToplam tutar: {toplam_tutar} TL")

if toplam_tutar > 500:
    indirimli_tutar = toplam_tutar * 0.90
    print("%10 indirim kazandınız!")
    print(f"Ödenecek indirimli tutar: {indirimli_tutar} TL")
else:
    print(f"Ödenecek tutar: {toplam_tutar} TL")
```
