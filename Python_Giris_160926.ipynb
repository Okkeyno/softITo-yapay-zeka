```python

# 3. DEĞİŞKENLER

# sehir adında bir değişken oluşturup içine yaşadığınız şehri atayın ve yazdırın.
sehir = "İstanbul"
print(sehir)

# puan adında bir değişkene 50 değerini verin, sonra değerini 75 olarak güncelleyip her iki değeri de yazdırın.
puan = 50
print("İlk Puan:", puan)

puan = 75
print("Güncellenmiş Puan:", puan)
```

```python
# 4. VERİ TİPLERİ
# Bir tam sayı, bir ondalıklı sayı, bir yazı ve bir boolean değişken oluşturup her birinin tipini type() ile yazdırın.
tam_sayi = 10
ondalikli_sayi = 3.14
yazi = "Merhaba"
durum = True

# Veri tiplerini type() ile yazdırma
print(type(tam_sayi))
print(type(yazi))
print(type(durum))

# 10 ve "10" değerlerinin tiplerini karşılaştırın — ikisi de aynı mı?
# İkisi aynı değil, biri integer, diğeri string
print("10 ve '10' aynı tipte mi?:", type(10) == type("10"))
```

```python
# 5. TEMEL MATEMATİK İŞLEMLERİ
x = 15
y = 4

print("Toplama:", x + y)
print("Çıkarma:", x - y)
print("Çarpma:", x * y)
print("Bölme:", x / y)

# Mod alma
print("17'nin 5'e bölümünden kalan:", 17 % 5)

# 2 üzeri 8
print("2'nin 8. kuvveti:", 2 ** 8)
```

```python
# 6. KULLANICIDAN VERİ ALMA (input)
# Kullanıcıdan adını alıp "Merhaba, [ad]!" şeklinde yazdıran bir program yazın.
ad = input("Lütfen adınızı girin: ")
print(f"Merhaba, [{ad}]!")

# . Kullanıcıdan yaşını alıp 10 yıl sonraki yaşını hesaplayıp yazdıran bir program yazın.
yas_str = input("Lütfen yaşınızı girin: ")
yas = int(yas_str)
print(f"10 yıl sonraki yaşınız: {yas + 10}")
```

```python
# 7. KARŞILAŞTIRMA OPERATÖRLERİ
a = 8
b = 12

print("a > b :", a > b)
print("a < b :", a < b)
print("a == b:", a == b)

# Kullanıcıdan bir sayı alıp bu sayının 100'e eşit olup olmadığını kontrol edin.
sayi = int(input("Bir sayı girin: "))
print("Girilen sayı 100'e eşit mi?:", sayi == 100)

```

```python

# 8. MANTIKSAL OPERATÖRLER
ogrenci_yas = 22
ogrenci_mi = False

#  yas = 22 ve ogrenci_mi = False değişkenleri için and, or ve not sonuçlarını yazdırın.
print((ogrenci_yas > 18) and ogrenci_mi)
print((ogrenci_yas > 18) or ogrenci_mi)
print(not ogrenci_mi)

# Bir kişinin hem 18 yaşından büyük hem de kayıtlı olup olmadığını kontrol eden bir ifade yazın.
kayitli_mi = True
kisi_yasi = 20
uygunluk = (kisi_yasi > 18) and kayitli_mi
print("Kişi hem 18'den büyük hem kayıtlı mı?:", uygunluk)
```

```python

# 9. BASİT IF-ELSE KULLANIMI
# Kullanıcıdan bir sayı alıp bu sayının pozitif mi negatif mi olduğunu yazdıran bir program yazın.
sayi_kontrol = int(input("Pozitif/Negatif kontrolü için bir sayı girin: "))
if sayi_kontrol > 0:
    print("Girilen sayı Pozitiftir.")
elif sayi_kontrol < 0:
    print("Girilen sayı Negatiftir.")
else:
    print("Girilen sayı Sıfırdır.")

# Kullanıcıdan not alıp 50 ve üzeri ise "Geçti", altında ise "Kaldı" yazdıran bir program yazın.
sinav_notu = float(input("Sınav notunuzu girin (0-100): "))
if sinav_notu >= 50:
    print("Geçti")
else:
    print("Kaldı")
```

```python
# 10. STRING (YAZI) İLE TEMEL İŞLEMLER
# Adınızı ve soyadınızı ayrı değişkenlerde tutup birleştirerek tam adınızı yazdırın.
ad = "Ali"
soyad = "Kaya"
tam_ad = ad + " " + soyad
print("Tam Ad:", tam_ad)

# Bir kelimenin kaç harften oluştuğunu len() ile bulun. Bir yazıyı tamamen büyük harfe, sonra tamamen küçük harfe çevirin.
kelime = "Programlama"
print(len(kelime))
print(kelime.upper())
print(kelime.lower())
```

```python
# 11. BASİT BİR PROGRAM ÖRNEĞİ
# Kullanıcıdan iki sayı alıp çarpımını yazdıran bir program yazın.
s1 = int(input("Çarpma işlemi için 1. sayıyı girin: "))
s2 = int(input("Çarpma işlemi için 2. sayıyı girin: "))
print(f"{s1} * {s2} = {s1 * s2}")

# Kullanıcıdan bir sayı alıp bu sayının karesini (kendisiyle çarpımını) hesaplayan bir program yazın.
sayi = int(input("Karesi alınacak sayıyı girin: "))
kare = sayi ** 2
print(kare)
```
