```python
def toplam_hesapla(*sayilar):  # * args
  toplam = 0
  for sayi in sayilar:
    toplam += sayi
  return toplam

print(toplam_hesapla(1,2,3,4,5))
print(toplam_hesapla(10,20,30))


def profil_olustur(**bilgiler): #** kwargs
  for anahtar, deger in bilgiler.items():
    print(f"{anahtar}:{deger}")

profil_olustur(isim="Berkay",yas=20,sehir="Ankara")
#
```

```python
class ogrenci:
  def __init__(self,isim,yas,notlar):
    self.isim = isim
    self.yas = yas
    self.notlar = notlar

  def ortalama_hesapla(self):
    return sum(self.notlar)/len(self.notlar)

  def durum(self):
    ort =self.ortalama_hesapla()
    return "Geçti" if ort >= 60 else "Kaldı"
```

```python
class Urun:
  def __init__(self,isim,fiyat,stok_adedi):
    self.isim = isim
    self.fiyat = fiyat
    self.stok_adedi = stok_adedi

  def toplam_hesapla(self):
    return self.fiyat * self.stok_adedi

  def stok_durumu(self):
    if self.stok_adedi == 0:
      return "Tükendi"
    elif self.stok_adedi < 10:
      return "Stok azalmış"
    else:
      return "Yeterli Stok"
```

```python
urun1 = Urun("Kularklık",150,8)
print(f"{urun1.isim} toplam değer : {urun1.toplam_hesapla()}")
print(f"{urun1.isim} stok durumu: {urun1.stok_durumu()}")
```
